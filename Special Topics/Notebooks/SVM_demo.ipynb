## Model selection 

### Linear kernel

```python
# linear SVM
C = 10
svm_linear = SVC(kernel = 'linear',C = C, random_state = 0)    # Create a SVC classifier using a linear kernel
svm_linear.fit(X_train.loc[:,['nPorosity','nAI']], y_train)    # Train the classifier
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Select the two features
features = ['nPorosity', 'nAI']
X = X_train[features].values
y = np.asarray(y_train)

# Create a mesh grid
x_min, x_max = X[:, 0].min() - 0.1, X[:, 0].max() + 0.1
y_min, y_max = X[:, 1].min() - 0.1, X[:, 1].max() + 0.1

xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300),np.linspace(y_min, y_max, 300))

grid = np.c_[xx.ravel(), yy.ravel()]

# Predict classes over the grid
Z = svm_linear.predict(grid).reshape(xx.shape)


# Decision function (for binary classification)
D = svm_linear.decision_function(grid).reshape(xx.shape)


plt.figure(figsize=(10, 7))

# Classification regions
plt.contourf(xx, yy, Z, alpha=0.2, cmap='coolwarm')

# Training observations
plt.scatter(X[:, 0], X[:, 1],c=y,cmap='coolwarm',edgecolors='k',s=40,alpha=0.8)

# Decision boundary and margins
plt.contour(xx, yy, D,levels=[-1, 0, 1],colors='black',linestyles=['--', '-', '--'],linewidths=[1.5, 2, 1.5]
)

# Support vectors
plt.scatter(svm_linear.support_vectors_[:, 0],svm_linear.support_vectors_[:, 1],s=80,facecolors='none',edgecolors='green',linewidths=1.0,
    label='Support Vectors'
)

# plt.scatter(svm_poly3_C1.support_vectors_[:, 0],svm_poly3_C1.support_vectors_[:, 1],s=80,facecolors='none',edgecolors='green',linewidths=1.0,
#     label='Support Vectors'
# )

plt.xlabel('Normalized Porosity')
plt.ylabel('Normalized Acoustic Impedance')
plt.title('Linear SVM - Decision Boundary and Support Vectors')
plt.legend()
plt.colorbar(label='Predicted Class')
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import (accuracy_score,classification_report,ConfusionMatrixDisplay)

features = ['nPorosity', 'nAI']

# Predictions using testing data
y_pred = svm_linear.predict(X_test[features])

# Accuracy
print("Testing Accuracy:", accuracy_score(y_test, y_pred))

# Classification metrics
print(classification_report(y_test, y_pred))

# Confusion matrix
ConfusionMatrixDisplay.from_predictions(y_test,y_pred,cmap='Blues')

plt.title('SVM - Testing Confusion Matrix')
plt.show()
```

### Tuning of polynomial kernel

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV, StratifiedShuffleSplit

# Define hyperparameter ranges
C_range = np.logspace(-2, 2, 5)
degree_range = [2, 3, 4, 5]

# Define parameter grid
param_grid = dict(C=C_range, degree=degree_range)

# Cross-validation
cv = StratifiedShuffleSplit(n_splits=5,test_size=0.3,random_state=4)

# Grid search with polynomial SVM
grid = GridSearchCV(SVC(kernel='poly', random_state=0),param_grid=param_grid,cv=cv,scoring='accuracy',n_jobs=-1)

# Train models
grid.fit(X_train[['nPorosity', 'nAI']], y_train)

# Extract validation accuracy
scores = grid.cv_results_['mean_test_score'].reshape(len(C_range), len(degree_range))

print("Best parameters:", grid.best_params_)
print("Best validation accuracy:", grid.best_score_)
```

```python
plt.figure(figsize=(8, 6))    
plt.imshow(scores, cmap=plt.cm.hot)
plt.xlabel('degree')
plt.ylabel('C')
plt.colorbar()
plt.xticks(np.arange(len(degree_range)), degree_range, rotation=45)
plt.yticks(np.arange(len(C_range)), C_range)
plt.title('Validation accuracy')
```
