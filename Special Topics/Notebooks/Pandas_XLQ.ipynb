Las estructuras de datos son formas de organizar y almacenar información para facilitar su acceso, manipulación y análisis. En Pandas, las dos estructuras principales son **`Series`** y **`DataFrame`**, diseñadas para trabajar con datos etiquetados de manera sencilla y eficiente.

# 2.1 Series

Una **`Series`** es una estructura de datos unidimensional (1D), similar a una columna de una tabla. Está formada por un conjunto de valores y un índice asociado a cada uno de ellos, lo que permite identificar y acceder a los datos mediante etiquetas.


*Sintaxis:*

pd.Series(data, index=None, dtype=None, name=None)

data: son los datos que se almacenarán.

index: etiquetas asociadas a cada dato. 

dtype: tipo de dato de los elementos.

name: nombre que tendrá la Series.

**EJEMPLO 1:**

```python
import pandas as pd

# Crear una Series
presion = pd.Series([2500, 'dos', 2380, 2300, 2210])

print(presion)
```

**EJEMPLO 2:**

```python
import pandas as pd

# Crear una Series
presion = pd.Series(
    [2500, 2450, 2380],
    index=["Pozo A", "Pozo B", "Pozo C"],
    dtype="float64",
    name="Presión"
)

print(presion)
```

# 2.2 DataFrame

Un **`DataFrame`** es una estructura de datos bidimensional (2D) organizada en filas y columnas, similar a una hoja de cálculo de Excel o a una tabla de una base de datos. Cada columna de un **`DataFrame`** puede considerarse una **`Series`** y puede contener diferentes tipos de datos, como números, texto, fechas o valores booleanos.

*Sintaxis*

pd.DataFrame(data, index=None, columns=None, dtype=None)

data: contiene los datos que formarán el DataFrame.

index: define las etiquetas de las filas.

columns: define o selecciona las columnas.

dtype: permite especificar el tipo de dato.

**EJEMPLO 1:**

```python
import pandas as pd
datos = {
    "Pozo": ["Pozo A", "Pozo B", "Pozo C", "Pozo D"],
    "Profundidad_m": [2500, 2800, 3100, 2750],
    "Presion_psi": [3200, 3500, 3900, 3400],
    "Produccion_bpd": [850, 720, 950, 680]
}

df = pd.DataFrame(datos)
df
```

**NOTA:** Algo importante es que cada columna de un DataFrame es una Series.

```python
df["Presion_psi"]
```

**EJEMPLO 2:**

```python
# DataFrame (En colab las listas siempre tieenen que ser de la misma dimensión)
import pandas as pd
data = {
    "nombre": ["Ana", "Luis", "Pedro"],
    "edad": [23, 30, 28],
    "ciudad": ["Madrid", "Bogotá", "CDMX"]
}

df = pd.DataFrame(data)
print(type(df))
df
```

La selección de datos permite extraer únicamente la información que se necesita de un DataFrame. En Pandas es posible seleccionar columnas, filas o valores específicos utilizando diferentes métodos.

# 4.1 Selección de columnas

Para seleccionar una columna de un DataFrame, se escribe el nombre de la columna entre corchetes.

*Sintaxis:*

DataFrame["columna"]

**EJEMPLO 1:**

```python
# 1 columna
pokemon[["Generation"]]
```

También es posible seleccionar varias columnas al mismo tiempo. Para ello se coloca una lista con los nombres de las columnas dentro de los corchetes.

*Sintaxis:*

DataFrame[["columna1", "columna2"]]

**EJEMPLO 2:**

```python
pokemon[["Generation", "Pokemon\nName"]]
```

# 4.2 Selección de filas

Las filas pueden seleccionarse utilizando diferentes métodos. Una forma sencilla es mediante un rango de posiciones utilizando corchetes.

*Sintaxis:*

DataFrame[inicio:fin]

**EJEMPLO:**

```python
pokemon[10:13]
```

# 4.3 Selección de filas por posición **`.iloc`**

El método **`.iloc`** permite seleccionar filas de un DataFrame de acuerdo con su posición numérica. Las posiciones comienzan desde 0.

*Sintaxis:*

DataFrame.iloc[inicio:fin]

**EJEMPLO:**

# 4.4 Selección de filas por índice **`.loc`**

El método **`.loc`** permite seleccionar datos utilizando principalmente las etiquetas del índice y los nombres de las columnas.

*Sintaxis:*

DataFrame.loc[[indice1, indice2, indice3]]

```python
# filas #por indice
pokemon_copia = pokemon.copy() # para generar una copia y no una referencia del mismo
pokemon_copia.loc[[0, 1 ,7]]
#pokemon_copia.index = pokemon_copia["name"] #cambio de indice
```

# 4.5 Restablecer el índice

El método **`.reset_index()`** permite restablecer el índice de un DataFrame, generando nuevamente un índice numérico consecutivo que comienza desde 0.

*Sintaxis:*

DataFrame.reset_index(drop=False)

* drop=False: conserva el índice anterior y lo agrega como una nueva columna.
* drop=True: elimina el índice anterior y crea un nuevo índice numérico.

**EJEMPLO:**

```python
# reset index
pokemon_copia = pokemon_copia.reset_index(drop=True)
pokemon_copia
```

```python
# columnas y filas
pokemon[["Generation", "Pokemon\nName"]].iloc[10:13]
```

El filtrado de datos permite seleccionar únicamente las filas de un DataFrame que cumplen una o varias condiciones específicas. En Pandas, las condiciones generan valores True o False, y se conservan únicamente las filas donde la condición es True.

# 5.1 Filtrar por un valor específico

Para seleccionar las filas que contienen un valor específico en una columna, se puede utilizar el operador de comparación **`==`**.

Sintaxis:

DataFrame[DataFrame["columna"] == valor]

**EJEMPLO 1:**

```python
# filtro por nombre
filtro_1 = pokemon[pokemon["Pokemon\nName"] == "Charmander"]
filtro_1
```

**NOTA:** En este caso, Pandas revisa la columna "Pokemon\nName	" y selecciona únicamente las filas donde el nombre sea exactamente "Charmander".

**EJEMPLO 2:**

```python
pokemon[pokemon["Experience\nValue"] >= 100]
```

**NOTA:**  Selecciona todos los Pokémon cuya Experiencia sea mayor o igual a 100.

También puedes utilizar otros operadores de comparación:

*==    Igual que*

*!=   Diferente de*

*>     Mayor que*

*<    Menor que*

*>=    Mayor o igual que*

*<=    Menor o igual que*

# 5.2 Filtrar por varios valores

El método  **`.isin()`** permite comprobar si los valores de una columna se encuentran dentro de un conjunto de valores especificados.

*Sintaxis:*

DataFrame[DataFrame["columna"].isin([valor1, valor2, valor3])]

**EJEMPLO:**

```python
# filtrar por nombre
filtro_1 = pokemon[pokemon["Pokemon\nName"].isin(["Charmeleon", "Charizard", "Mew"])]
filtro_1
```

# 5.3 Filtrar con múltiples condiciones

Pandas también combina dos o más condiciones para realizar filtros más específicos.

Los operadores:

* & → significa Y: ambas condiciones deben cumplirse.
* | → significa O: al menos una condición debe cumplirse.

*Sintaxis con &:*

DataFrame[    (condicion1) & (condicion2)]

**EJEMPLO 1:**

```python
# filtrar por rangos "| significa o" "& significa y"
filtro_1 = pokemon[(pokemon["Base Friendship"] >= 100) & (pokemon["Experience\nValue"] >= 100)]
filtro_1
```

**NOTA:** En este caso, únicamente se seleccionan los Pokémon que cumplen las dos condiciones simultáneamente.

Si queremos que se cumpla una condición o la otra, utilizamos |.
    
*Sintaxis con | :*

DataFrame[    (condicion1) | (condicion2)]

**EJEMPLO 2:**

```python
filtro_2 = pokemon[
    (pokemon["Base Friendship"] >= 100) | (pokemon["Experience\nValue"] >= 200)]

filtro_2
```

**NOTA:** No es necesario que ambas condiciones se cumplan al mismo tiempo.

# 6.1 Creación de columnas

Para crear una nueva columna se escribe el nombre de la columna entre corchetes y se asignan los valores que contendrá.

*Sintaxis:*

* DataFrame["nueva_columna"] = (DataFrame["columna1"] / DataFrame["columna2"])

```python
# creación de columnas
pokemon["Generacion_Aumento"] = pokemon_copia["Generation"] + 2

pokemon.head() ## ¿Recuerdas que hace .head()?
```

# 6.2 Insertar una columna

El método **`.insert()`** permite agregar una nueva columna en una posición específica del DataFrame.

*Sintaxis:*

DataFrame.insert(posicion, "nombre_columna", valores)

# 6.3 Eliminar columnas

El método **`.drop()`** permite eliminar una o varias columnas de un DataFrame.

*Sintaxis:*

DataFrame.drop(columns=["columna1", "columna2"])

**EJEMPLO:**


```python
# Eliminar columna
pokemon.drop(columns = ["Amistad_","Generacion_Aumento"], inplace = True) # inplace modifica directamente
pokemon
```

**NOTA:** **`inplace=True`** indica que la modificación se realizará directamente sobre el DataFrame original.

# 6.4 Agregar una fila al final

Para agregar una nueva fila al final de un DataFrame, se puede utilizar **`.loc`** junto con **`len()`**. La función **`len()`** obtiene el número de filas actuales y permite utilizar la siguiente posición disponible.

*Sintaxis:*

DataFrame.loc[len(DataFrame)] = { "columna1": valor1,"columna2": valor2}

**EJEMPLO:**


```python
# agregar filas al final
pokemon.loc[len(pokemon)] = {"pokedex_id": 252,
                             "name": "Nuevo2",
                             "height": 23,
                             "weight": 30,
                             "base_experience": 60,
                             "type_1": "fire",
                             "type_2": "flying",
                             "hp": 50,
                             "attack": 50,
                             "defense": 50,
                             "special_attack": 100,
                             "special_defense": 60,
                             "speed": 30}

pokemon
```

# 6.5 Agregar varias filas

Cuando se desean agregar varias filas, primero se puede crear un nuevo DataFrame con los registros y posteriormente unirlo al DataFrame original mediante **`pd.concat()`**.

*Sintaxis:*

nuevo_df = pd.DataFrame(datos)

DataFrame = pd.concat( [DataFrame, nuevo_df], ignore_index=True)

**EJEMPLO:** Se usara un nuevo archivo tipo **`.csv`**

```python
pokemon2 = pd.read_csv("pokemon_stats_2025_1.csv")
pokemon2.head()
```

```python
# agregar filas como dataframe
filas = pd.DataFrame({"pokedex_id": [253, 254],
                             "name": ["Nuevo3", "Nuevo4"],
                             "height": [23, 21],
                             "weight": [30, 40],
                             "base_experience": [60, 50],
                             "type_1": ["fire", "rock"],
                             "type_2": ["flying", "dark"],
                             "hp": [50, 40],
                             "attack": [50, 20],
                             "defense": [50, 35],
                             "special_attack": [100, 32],
                             "special_defense": [60, 25],
                             "speed": [30, 23]})

pokemon2 = pd.concat([pokemon2, filas], ignore_index=True) #ignore_idex hace que el indice se regenere
pokemon2
```

# 6.6 Insertar filas en una posición específica

**`pd.concat()`** también permite insertar nuevas filas en una determinada posición. Para hacerlo, se divide el DataFrame utilizando .iloc y se colocan las nuevas filas entre ambas partes.

Sintaxis:

DataFrame = pd.concat([DataFrame.iloc[:posicion],nuevas_filas,DataFrame.iloc[posicion:]],ignore_index=True)

**EJEMPLO:**

```python
# insertar nuevas filas como dataframe
filas = pd.DataFrame({"pokedex_id": [255, 256],
                             "name": ["Nuevo5", "Nuevo6"],
                             "height": [23, 21],
                             "weight": [30, 40],
                             "base_experience": [60, 50],
                             "type_1": ["fire", "rock"],
                             "type_2": ["flying", "dark"],
                             "hp": [50, 40],
                             "attack": [50, 20],
                             "defense": [50, 35],
                             "special_attack": [100, 32],
                             "special_defense": [60, 25],
                             "speed": [30, 23]})


pokemon2 = pd.concat([pokemon2.iloc[:2], filas, pokemon2.iloc[2:]], ignore_index=True)
pokemon2
```

DataFrame.drop(index=[indice1, indice2])

# 6.9 Eliminar filas con múltiples condiciones

Los filtros también pueden utilizarse para eliminar registros que cumplan condiciones numéricas.
F
*Sintaxis:*

filtro = DataFrame[(DataFrame["columna1"] >= valor) |(DataFrame["columna2"] >= valor)]

# 6.15 Reemplazar valores

El método **`.replace()`** permite sustituir uno o varios valores existentes por nuevos valores.

*Sintaxis:*

DataFrame["columna"].replace([valor1, valor2],[nuevo1, nuevo2])

**EJEMPLO:**

```python
# remplazar valores
pokemon2["nombre"] = pokemon2["nombre"].replace(["Charizard", "Weedle"], ["CHARIZARD", "WEEDLE"])
pokemon2.head(7)
```

# 6.16 Reemplazar valores utilizando un diccionario

También se puede utilizar un diccionario para indicar qué valor será reemplazado y cuál será su sustituto.

*Sintaxis:*

DataFrame["columna"].replace({valor_anterior: valor_nuevo})

**EJEMPLO:**

```python
# remplazar valores con diccionarios
pokemon2["nombre"] = pokemon2["nombre"].replace({"CHARIZARD": "Charizard", "WEEDLE": "Weedle"})
pokemon2.head(7)
```

**NOTA:** **`.replace()`** permite sustituir valores utilizando diferentes estructuras. Con listas, los valores se relacionan según su posición; con un diccionario, se establece explícitamente la relación valor_anterior: valor_nuevo. El uso de diccionarios suele ser más claro cuando se realizan múltiples reemplazos.

El ordenamiento de datos permite organizar las filas de un DataFrame para facilitar su visualización, comparación y análisis. Los datos pueden ordenarse de forma ascendente o descendente, ya sea utilizando los valores de una o varias columnas o las etiquetas del índice.

Para realizar estas operaciones, Pandas proporciona principalmente los métodos **`.sort_values()`** y **`.sort_index()`**.

## 7.1 Ordenar por valores

El método  **`.sort_values()`** permite ordenar las filas de un DataFrame de acuerdo con los valores de una o varias columnas. Los datos pueden organizarse en orden ascendente o descendente.

*Sintaxis 1:*

DataFrame.sort_values(by="columna",ascending=True)

* ascending=True: ordena el índice de forma ascendente

**EJEMPLO:**


```python
# ordenar en orden descendente
pokemon2.sort_values("attack", ascending = False)
```

# <font color="blue">**10. Unión de DataFrames** </font>

</td>
</tr>
</table>

La unión de DataFrames permite combinar información proveniente de dos conjuntos de datos diferentes utilizando una columna en común. Esta operación es útil cuando la información relacionada se encuentra distribuida en distintos archivos o tablas.

La función **`pd.merge()`** permite unir dos DataFrames utilizando los valores coincidentes de una columna en común.

*Sintaxis:*

pd.merge( DataFrame1,DataFrame2,on="columna_comun")

**EJEMPLO:**

Primero se carga el segundo archivo que contiene la información adicional.

```python
# datos para merge
region = pd.read_csv("pokemon_stats_2025_2.csv")
region
```

Para realizar la unión, ambos DataFrames deben tener una columna que permita relacionar la información. La columna en común será **`name`**

```python
# merge
pokemon2.rename(columns = {"nombre": "name"}, inplace = True) # cambiando nombre de columna
pokemon_region = pd.merge(pokemon2, region, on = "name")
pokemon_region
```

**NOTA:** Entonces "name" funciona como la columna de unión: Pandas busca los nombres que coinciden en pokemon y region y reúne la información de ambos DataFrames en uno nuevo.
