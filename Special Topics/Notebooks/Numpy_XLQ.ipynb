## **2.1 Arreglos**
Antes de comenzar a utilizar NumPy, es importante comprender qué es un arreglo (array), ya que constituye la estructura fundamental con la que trabaja esta biblioteca.

Un arreglo es una colección ordenada de elementos almacenados bajo un mismo nombre. Todos los elementos ocupan una posición específica dentro del arreglo, la cual puede identificarse mediante un índice.

Los arreglos permiten:

* Almacenar grandes cantidades de datos de manera organizada.
* Acceder rápidamente a cualquier elemento mediante su posición.
* Realizar operaciones matemáticas sobre todos los elementos de forma simultánea.

En Python, los arreglos numéricos suelen implementarse mediante la biblioteca NumPy, que está optimizada para cálculos científicos y de ingeniería.

Antes de utilizar NumPy, es necesario importar la biblioteca en nuestro programa. Python no tiene disponibles directamente las funciones especializadas de NumPy para trabajar con arreglos, matrices y operaciones numéricas.

Se importar NumPy de esta manera:

```python
import numpy as np
```

**NOTA:** La palabra **`as`** permite asignarle un alias a la biblioteca. En este caso, se utiliza **`np`** porque es la forma más común de abreviar NumPy y permite escribir un código más corto y fácil de entender.

**EJEMPLO**:

Supongamos que tenemos las calificaciones de cinco estudiantes. Sin utilizar un arreglo, tendríamos que almacenar cada calificación en una variable diferente:

```python
calificacion_1 = 8.5
calificacion_2 = 9.0
calificacion_3 = 7.5
calificacion_4 = 10.0
calificacion_5 = 8.0
```

Con NumPy, podemos almacenar todas las calificaciones dentro de un solo arreglo:

```python
calificaciones = np.array([8.5, 9.0, 7.5, 10.0, 8.0])

print(calificaciones)
```

De esta manera, todas las calificaciones se encuentran organizadas en una sola estructura y podemos realizar operaciones sobre ellas fácilmente. Por ejemplo, podemos calcular el **promedio del grupo**:

```python
promedio = np.mean(calificaciones)

print("Promedio del grupo:", promedio)
```

## **2.2 Vectores**

Un vector es un arreglo de una sola dimensión (1D) formado por una secuencia ordenada de elementos. En NumPy, los vectores permiten almacenar varios valores dentro de una misma estructura y acceder a cada uno de ellos mediante su posición o índice.

Los vectores se utilizan para representar:

* Coordenadas en el espacio.
* Mediciones experimentales.
* Series de datos.
* Características de un objeto en aprendizaje automático.

**EJEMPLO:**

Supongamos que queremos almacenar las temperaturas registradas durante siete días. En lugar de crear una variable para cada día, podemos representar todas las mediciones mediante un vector:

```python
temperaturas = np.array([22, 24, 23, 25, 27, 26, 24])

print(temperaturas)
```

En este caso,**`temperaturas`** es un vector de siete elementos, donde cada posición corresponde a la temperatura registrada en un día de la semana.

## **2.3 Matrices**

Una matriz es un arreglo de dos dimensiones (2D) en el que los elementos se encuentran organizados en filas y columnas. En NumPy, las matrices permiten almacenar y trabajar con conjuntos de datos relacionados dentro de una misma estructura.

Las matrices se utilizan para representar diferentes tipos de información, por ejemplo:

* Tablas de datos numéricos.
* Mediciones de diferentes variables.
* Datos experimentales.
* Sistemas de ecuaciones.
* Información de diferentes muestras u observaciones.

**EJEMPLO:**

Supongamos que queremos almacenar las calificaciones de cuatro estudiantes en tres materias diferentes. Podemos organizar esta información mediante una matriz, donde cada fila representa a un estudiante y cada columna representa una materia:

```python
calificaciones = np.array([
    [8, 9, 10],
    [7, 8, 9],
    [9, 10, 9],
    [8, 7, 10]
])

print(calificaciones)
```

En este caso, **`calificaciones`** es una matriz de 4 filas y 3 columnas. Es decir, contiene las calificaciones de 4 estudiantes en 3 materias.

# <font color="blue">**3. Creación de arreglos**</font>

En NumPy, un arreglo puede crearse de diferentes maneras dependiendo de los datos que tengamos disponibles y del tipo de arreglo que necesitemos. No siempre será necesario escribir manualmente todos los valores que formarán parte del arreglo, ya que NumPy proporciona funciones que permiten **generarlos automáticamente**.

Por ejemplo, podemos crear un arreglo a partir de una lista de datos que ya conocemos, como las temperaturas registradas durante una semana. También podemos crear arreglos de un tamaño determinado y llenarlos automáticamente con ceros, unos o cualquier otro valor.

Además, NumPy permite generar secuencias de números indicando un intervalo o la cantidad de valores que queremos obtener. De la misma manera, podemos crear arreglos con números aleatorios, lo cual resulta especialmente útil para realizar simulaciones, generar datos de prueba o trabajar con métodos numéricos.

Por esta razón, NumPy cuenta con distintas funciones para la creación de arreglos, entre las que se encuentran np.array(), np.zeros(), np.ones(), np.eye(), np.full(), np.arange(), np.random.rand(), np.random.randint() y np.linspace().  

* **np.array()**
  
La función **`np.array()`** se utiliza para crear arreglos en NumPy. Un arreglo es una estructura que permite almacenar y organizar varios valores dentro de una misma variable, facilitando el manejo y las operaciones con los datos.

La dimensión del arreglo dependerá de cómo estén organizados los datos que introduzcamos.

Si utilizamos una lista simple, NumPy crea un arreglo de una dimensión (1D), es decir, un vector.

*Sintaxis básica:*

np.array(datos)

**EJEMPLO:** Podemos almacenar las temperaturas registradas durante cinco días

```python
temperaturas = np.array([22, 24, 23, 26, 25])

print(temperaturas)
```

En cambio, si organizamos los datos en filas y columnas, NumPy puede crear un arreglo de dos dimensiones (2D), es decir, una matriz.

**Ejemplo:** Podemos representar las calificaciones de tres estudiantes en tres materias

```python
calificaciones = np.array([
    [8, 9, 10],
    [7, 8, 9],
    [9, 10, 8]
])

print(calificaciones)
```

* **np.zeros()**

La función **`np.zeros()`** crea un arreglo en el que todos los elementos tienen un valor inicial de cero. Las dimensiones del arreglo se especifican dentro de la función.

*Sintaxis básica:*

np.zeros(dimensiones)

**Ejemplo**: Para crear una matriz de 3 filas y 4 columnas llena de ceros:

```python
ceros = np.zeros((3, 4))

print(ceros)
```

**NOTA:**
Esta función es útil cuando necesitamos inicializar una matriz para posteriormente almacenar o calcular nuevos valores.

* **np.ones()**
  
La función **`np.ones()`** funciona de manera similar a **`np.zeros()`**, pero crea un arreglo en el que todos los elementos tienen el valor 1.

*Sintaxis básica:*

np.ones(dimensiones)

**Ejemplo:** Para crear una matriz de 2 filas y 4 columnas llenas del valor de 1. 

```python
unos = np.ones((2, 4))

print(unos)
```

* **np.full()**

La función **`np.full()`** permite crear un arreglo con determinadas dimensiones y llenar todos sus elementos con un valor específico.

*Sintaxis básica:*

np.full(dimensiones, valor)

**Ejemplo:** Supongamos que queremos crear una matriz de 2 filas y 3 columnas donde todos los valores iniciales sean 10:

```python
matriz = np.full((2, 3), 10)

print(matriz)
```

* **np.eye()**

La función **`np.eye()`** permite crear una matriz cuyos elementos son ceros, excepto aquellos que se encuentran en una diagonal determinada, los cuales tienen el valor 1.

*Sintaxis básica:*

np.eye(n)

Donde **`n`** representa el número de filas y columnas de la matriz cuadrada.

**Ejemplo:** Podemos crear una matriz identidad de 3 × 3:

```python
identidad = np.eye(3)

print(identidad)
```

Una característica importante de **`np.eye()`** es que no está limitada a matrices cuadradas. Podemos indicar por separado el número de filas y columnas.

**Ejemplo:** Para crear una matriz de 3 filas y 5 columnas:

```python
matriz = np.eye(3, 5)

print(matriz)
```

* **np.arange()**

La función **`np.arange()`** permite crear un arreglo con una secuencia de valores separados por un determinado intervalo o paso. 

*Sintaxis básica:*

np.arange(inicio, fin, paso)

**Ejemplo:** Queremos generar números desde 0 hasta 20 avanzando de 2 en 2:

```python
numeros = np.arange(0, 20, 2)

print(numeros)
```

* **np.linspace()**

La función **`np.linspace()`** crea un arreglo con una cantidad determinada de valores igualmente espaciados entre un valor inicial y uno final.

*Sintaxis básica:*

np.linspace(inicio, fin, cantidad)

Donde **`cantidad`** es el número de valores que queremos generar.

**Ejemplo:** Queremos generar cinco valores igualmente espaciados entre 0 y 10:

```python
valores = np.linspace(0, 10, 5)

print(valores)
```

**NOTA:**
Una diferencia importante con **`np.arange()`** es que aquí indicamos cuántos valores queremos obtener, en lugar de indicar directamente el tamaño del paso. Además, de manera predeterminada, **`linspace()`** sí incluye el valor final.

* **np.random.rand()**

La función **`np.random.rand()`** permite generar arreglos de números pseudoaleatorios de punto flotante distribuidos uniformemente entre 0 y 1 (incluye 0 pero no 1).

*Sintaxis básica:*

np.random.rand(filas, columnas)

**Ejemplo:** Para generar una matriz de 2 filas y 3 columnas:

```python
aleatorios = np.random.rand(2, 3)

print(aleatorios)
```

* **np.random.randint()**

La función **`np.random.randint()`** genera números enteros pseudoaleatorios dentro de un intervalo determinado.

*Sintaxis básica:*

np.random.randint(inicio, fin, size)

**Ejemplo:** Supongamos que queremos generar cinco calificaciones enteras aleatorias entre 5 y 10:

```python
calificaciones = np.random.randint(5, 11, size=5)

print(calificaciones)
```

En NumPy, las dimensiones indican la forma en que los datos están organizados dentro de un arreglo. Dependiendo de cómo se agrupen los elementos, un arreglo puede tener una, dos, tres o más dimensiones.

Cada dimensión puede entenderse como un nivel de organización de los datos. Por ejemplo, un solo valor corresponde a un arreglo de cero dimensiones (0D), una secuencia de valores forma un arreglo de una dimensión (1D) y una estructura organizada en filas y columnas corresponde a un arreglo de dos dimensiones (2D).

Podemos visualizarlo de la siguiente manera:

- 0D → Escalar: contiene un único valor, como el número `5`.
- 1D → Vector: contiene una secuencia de valores.
- 2D → Matriz: contiene valores organizados en filas y columnas.
- 3D → Arreglo tridimensional: puede visualizarse como un conjunto de matrices.

Conocer las dimensiones y características de un arreglo es importante para saber cómo están organizados los datos y cómo podemos trabajar con ellos. Para obtener esta información, NumPy proporciona atributos como `.ndim`, `.shape`, `.size` y `.dtype`.

* **.ndim**

El atributo  `.ndim ` permite conocer el número de dimensiones o ejes que tiene un arreglo.

*Sintaxis básica:*

arreglo.ndim

**EJEMPLO:**

```python
matriz = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
print("MATRIZ")
print(matriz.ndim)

vector = np.array([10, 20, 30, 40])
print("VECTOR")
print(vector.ndim)
```

**NOTA:**

El resultado es `2` porque `matriz` está organizado mediante filas y columnas, por lo que se trata de un arreglo bidimensional.

El resultado es `1` porque `vector` tiene una sola dimensión.


* **.shape**

El atributo `.shape` permite conocer cómo están distribuidos los elementos en cada dimensión del arreglo.

*Sintaxis básica:*

arreglo.shape

**EJEMPLO:**

```python
matriz = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(matriz.shape)
```

**NOTA:**
El resultado (2, 3) significa que el arreglo tiene 2 filas y 3 columnas

* **.size** 
    
El atributo `.size` indica la cantidad total de elementos almacenados dentro de un arreglo, independientemente de cómo estén distribuidos.

*Sintaxis básica:*

arreglo.size

**EJEMPLO:**

```python
matriz = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
print(matriz.size)
```

**NOTA:**

La respuesta muestra que tiene 6 elementos en la matriz.

El atributo `.size` es util cuando se quiere saber cuántos datos contiene un arreglo.


* **.dtype** 

El atributo .dtype permite conocer el tipo de dato que NumPy utiliza para almacenar los elementos de un arreglo.

*Sintaxis básica:*

arreglo.dtype

**EJEMPLO:**

```python
import numpy as np
numeros = np.array([10, 20, 30, 40])

print(numeros.dtype)
```

**NOTA:**

El resultado `int64` indica que los elementos están almacenados como números enteros.

```python
temperaturas = np.array([22.5, 23.8, 25.1, 24.6])

print(temperaturas.dtype)
```

**NOTA:**

El resultado `float64` indica que los elementos están almacenados son números decimales.

En NumPy, los datos de un arreglo pueden reorganizarse o cambiar su número de dimensiones sin necesidad de introducir nuevamente sus valores. Esto es útil cuando necesitamos adaptar la estructura de los datos para realizar cálculos, trabajar con matrices o preparar información para otros procesos.

NumPy proporciona diferentes herramientas para modificar la estructura de los arreglos, entre ellas `reshape()`, `flatten()`, `ravel()`, `transpose()`, `expand_dims()` y `squeeze()`.

* **reshape()**
  
El método `reshape()` permite cambiar la forma de un arreglo, es decir, reorganizar sus elementos en un número diferente de filas, columnas o dimensiones.

*Sintaxis básica:*

arreglo.reshape(filas, columnas)

**EJEMPLO: Al inicio se tiene un vector de 6 elementos**

```python
numeros = np.array([1, 2, 3, 4, 5, 6])

print(numeros)
print(numeros.shape)
```

**Podemos convertirlo en una matriz de 2 filas y 3 columnas:**

```python
matriz = numeros.reshape(2, 3)

print(matriz)

print(matriz.shape)
```

**NOTA:** La nueva forma también debe contener 6 elementos. Por ejemplo, 2 × 3 = 6 o 3 × 2 = 6.

* **flatten()**

El método `flatten()` permite convertir un arreglo de varias dimensiones en un arreglo de una sola dimensión (1D). Utilizado para convertir una matriz en un vector.

*Sintaxis básica:*

arreglo.flatten()

**EJEMPLO: La matriz de dimensiones (2, 3) se va a convertir en un vector de 6 elementos.**

```python
matriz = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(matriz)
```

**`flatten()` va a crear una copia independiente de los datos. Por lo tanto, modificar el nuevo vector no modifica el arreglo original.**

```python
vector = matriz.flatten()

print(vector)
```

* **ravel()**
  
El método `ravel()` también permite convertir un arreglo multidimensional en un arreglo de una sola dimensión.

La diferencia principal entre `flatten()` y `ravel()` está en cómo manejan los datos en la memoria
.
En  `flatten()` la matriz original no cambia porque creo una copia independiente.

En `ravel()`evita crear una copia de los datos. Esto significa que el nuevo arreglo y el original pueden compartir la misma información en memoria.

*Sintaxis básica:*

arreglo.ravel()

**EJEMPLO:**

```python
matriz = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

vector = matriz.ravel()

print(vector)
```

**Al modificar un valor en la posición 0 del vector, también se modifica en la matriz porque comparten los mismos datos en memoria.**

```python
vector[0] = 100

print(vector)
print(matriz)
```

* **transpose()**

El método `transpose()` permite cambiar la posición de los datos de una matriz, haciendo que las filas pasen a ser columnas y las columnas pasen a ser filas.

*Sintaxis básica:*

* arreglo.transpose()

* arreglo.T

**EJEMPLO: Tenemos una matriz de 2 filas y 3 columnas:**

```python
matriz = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(matriz)
```

**Al aplicar la transposición la matriz va a pasar de
2 filas × 3 columnas a 3 filas × 2 columnas**

```python
transpuesta = matriz.transpose()

print(transpuesta)
```

* **np.expand_dims()**

La función `np.expand_dims()` permite agregar una nueva dimensión a un arreglo.

A diferencia de `reshape()`, aquí indicamos principalmente en qué posición queremos agregar el nuevo eje mediante el parámetro `axis`.

*Sintaxis básica:*

np.expand_dims(arreglo, axis)

**EJEMPLO:**

```python
vector = np.array([10, 20, 30])
print(vector)

print(vector.shape)
```

**Se agrega una nueva dimensión**

```python
nuevoD = np.expand_dims(vector, axis=1)

print(nuevoD)
print(nuevoD.shape)
```

* **np.squeeze()**

La función `.squeeze()` permite eliminar las dimensiones innecesarias de un arreglo, específicamente aquellas que contienen un solo elemento pero no elimina datos. Los valores del arreglo permanecen iguales; lo único que cambia es la forma `.shape` en la que están organizados.

*Sintaxis básica:*

np.squeeze(arreglo)

**EJEMPLO 1: Se un arreglo con 3 columnas y 1 fila**

```python
datos = np.array([[10, 20, 30]])

print(datos)
print(datos.shape)
```

**Eliminamos la dimensión de tamaño 1**

```python
nuevo = datos.squeeze()

print(nuevo)
print(nuevo.shape)
```

**EJEMPLO 2: Se un arreglo con 3 filas y 1 columna:**

```python
import numpy as np

datos = np.array([
    [10],
    [20],
    [30]
])

print(datos)
print(datos.shape)
```

**Eliminamos la dimensión de tamaño 1**

```python
nuevo = np.squeeze(datos)

print(nuevo)
print(nuevo.shape)
```

**NOTA: `np.squeeze()` elimina dimensiones de tamaño 1, pero nunca elimina los datos que contiene el arreglo.**

Una de las principales ventajas de NumPy es que permite realizar operaciones matemáticas directamente sobre los elementos de un arreglo, sin necesidad de recorrerlos individualmente mediante ciclos.

## **6.1 Operaciones aritméticas**

**Suma:**           + o `np.add()`

**Resta:**          - o `np.subtract()`

**Multiplicación:** * o `np.multiply()`

**División**        / o `np.divide()`

**Potencia:**       ** o `np.power()`

**EJEMPLOS:**

```python
a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Suma:")
print(a + b)
print("Resta:")
print(a - b)
print("Multiplicación:")
print(a * b)
print("División:")
print(a / b)
```

## **6.2 Funciones matemáticas**

**Raíz cuadrada:** `np.sqrt()`

**Exponencial:** `np.exp()`

**Logaritmo natural:** `np.log()`

**Valor absoluto:** `np.abs() `

**EJEMPLOS:**

```python
numeros = np.array([1, 4, 9, 16])

print("Raíz cuadrada:", np.sqrt(numeros))

print("Exponencial:", np.exp(numeros))

print("Logaritmo:", np.log(numeros))

print("Valor absoluto:", np.abs(numeros))
```

**Seno:** `np.sin()`  
**Coseno:** `np.cos()`

**EJEMPLOS:**

```python
angulos = np.array([0, 30, 45, 90])
radianes = np.deg2rad(angulos)

print("Seno:", np.sin(radianes))

print("Coseno:", np.cos(radianes))
```

## **6.3 Operaciones por ejes**

Cuando trabajamos con una matriz, podemos realizar una operación sobre todos los datos, o podemos pedirle a NumPy que la realice separadamente según sus filas o columnas. Para ello se utiliza el parámetro `axis`.

En una matriz de dos dimensiones:

* `axis=0` → obtiene un resultado para cada columna.
* `axis=1` → obtiene un resultado para cada fila.

**EJEMPLO:Se tienen las calificaciones de 3 estudiantes en 3 materias.**

**Si queremos conocer el promedio de cada materia, necesitamos trabajar por columnas:**


```python
calificaciones = np.array([
    [8,  9, 10],
    [7,  8,  9],
    [9, 10,  8]
])

print("Operación por columnas: axis=0")
np.mean(calificaciones, axis=0)
```

**EJEMPLO:**

**Si queremos conocer el promedio de cada estudiante. Como cada estudiante corresponde a una fila:**

NumPy proporciona diferentes funciones que permiten realizar cálculos estadísticos sobre los datos almacenados en un arreglo. Estas herramientas son útiles para resumir y analizar conjuntos de datos, permitiendo identificar valores representativos, conocer su dispersión y determinar cómo se distribuyen.

Entre las principales operaciones estadísticas se encuentran la `media`, `mediana`, `desviación estándar`, `varianza`, `valores mínimo` y `máximo`y `percentiles`.


## **8.1 Media**

La media aritmética `np.mean()`, también conocida como promedio, representa un valor central de un conjunto de datos. Se obtiene sumando todos los valores y dividiendo el resultado entre el número total de elementos.

**EJEMPLO:**

```python
datos = np.array([10, 12, 14, 16, 18])

media = np.mean(datos)

print("Media:", media)
```

## **8.2 Mediana**

La mediana `np.median()` es el valor que se encuentra en el centro de un conjunto de datos después de ordenarlos de menor a mayor. Si existe un número impar de elementos, corresponde al valor central. Si existe un número par, se obtiene el promedio de los dos valores centrales.

**EJEMPLO:**

```python
mediana = np.median(datos)

print("Mediana:", mediana)
```

## **8.3 Desviación estándar**

La desviación estándar `np.std()` indica qué tan dispersos o alejados se encuentran los datos respecto a la media. Una desviación estándar pequeña indica que los valores se encuentran relativamente cerca de la media, mientras que una desviación estándar grande indica una mayor dispersión.

**EJEMPLO:**

```python
desviacion = np.std(datos)

print("Desviación estándar:", desviacion)
```

## **8.4 Varianza**

La varianza `np.var()` también mide la dispersión de los datos respecto a la media. Se obtiene a partir de las diferencias entre cada valor y la media elevadas al cuadrado.

**EJEMPLO:**

```python
varianza = np.var(datos)

print("Varianza:", varianza)
```

## **8.5 Valor mínimo**

La función `np.min()` permite encontrar el valor más pequeño contenido en un arreglo.

**EJEMPLO:**

```python
minimo = np.min(datos)

print("Valor mínimo:", minimo)
```

## **8.6 Valor máximo**

La función `np.max()` permite encontrar el valor más grande contenido en un arreglo.

**EJEMPLO:**

```python
maximo = np.max(datos)

print("Valor máximo:", maximo)
```

## **8.7 Percentiles**

Los percentiles `np.percentile()` son valores que permiten dividir un conjunto de datos ordenados en 100 partes. Se utilizan para conocer cómo se distribuyen los datos y determinar qué porcentaje de las observaciones se encuentra en o por debajo de cierto valor.

**EJEMPLO:**

```python
datos = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90])

print(np.percentile(datos, 25))
print(np.percentile(datos, 50))
print(np.percentile(datos, 75))
```

Los resultados significan:

P25 = `30` → aproximadamente el 25 % de los datos está en o por debajo de `30`.

P50 = `50` → aproximadamente el 50 % de los datos está en o por debajo de `50`.

P75 = `70` → aproximadamente el 75 % de los datos está en o por debajo de `70`.

</tr>

El álgebra lineal es una rama de las matemáticas que trabaja principalmente con vectores, matrices y sistemas de ecuaciones. NumPy proporciona herramientas que permiten realizar estas operaciones de forma directa mediante funciones como `Producto punto`, `Multiplicación matricial`, `Determinante`, `Matriz inversa` y `Resolver sistemas de ecuaciones`.


## **9.1 Producto punto**

El producto punto o producto escalar `np.dot()` es una operación entre dos vectores del mismo tamaño. Consiste en multiplicar los elementos que ocupan la misma posición y después sumar los resultados.

**EJEMPLO:**

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

resultado = np.dot(a, b)

print("Producto punto:", resultado)
```

## **9.2 Multiplicación matricial**

La multiplicación matricial `np.matmul()` permite multiplicar dos matrices siguiendo una regla diferente a la multiplicación elemento por elemento. Para obtener cada elemento de la matriz resultante, se toma una fila de la primera matriz y una columna de la segunda, se multiplican sus elementos correspondientes y finalmente se suman.

**EJEMPLO:**

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

resultado = np.matmul(A, B)
print(resultado)
```

## **9.3 Determinante**

El determinante `np.linalg.det()` es un único valor numérico que se obtiene a partir de una matriz cuadrada, es decir, una matriz que tiene el mismo número de filas y columnas.

**EJEMPLO:**

```python
A = np.array([
    [1, 2],
    [3, 4]
])

determinante = np.linalg.det(A)

print("Determinante:", determinante)
```

## **9.4 Matriz inversa**

La matriz inversa `np.linalg.inv()` es una matriz asociada a otra matriz cuadrada.

La matriz identidad funciona de manera parecida al número 1 en una multiplicación común.

**EJEMPLO:**

```python
A = np.array([
    [1, 2],
    [3, 4]
])

inversa = np.linalg.inv(A)

print(inversa)
```

**Su característica principal es que, cuando una matriz A se multiplica por su matriz inversa, el resultado es la matriz identidad.**

```python
Identidad= np.matmul(A, inversa)

print(np.round(Identidad))
```

**NOTA:** `np.round()` es para redondear el resultado.

## **9.5 Resolver sistemas de ecuaciones**

En NumPy también permite resolver sistemas de ecuaciones lineales con `np.linalg.solve()`.

**EJEMPLO:**

**Se tiene el siguiente sistema de ecuaciones:**
$$
2x+y=5
$$
$$
x−y=1
$$
**Este sistema puede representarse en forma matricial mediante:**
$$
Ax=b
$$
**En la matriz A contiene los coeficientes de las incógnitas y el vector B contiene los resultados de las ecuaciones.**

```python
A = np.array([
    [2,  1],
    [1, -1]
])

b = np.array([5, 1])

solucion = np.linalg.solve(A, b)

print(solucion)
```
